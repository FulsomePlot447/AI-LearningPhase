```python
import torch
import random
import numpy as np
```

```python
class linear():
    def __init__(self, in_feature: int, out_feature: int):

        self.in_feature = in_feature
        self.out_feature = out_feature
        self.weights = np.random.randn(in_feature, out_feature) * 0.01
        self.bias = np.zeros(out_feature)
        self.x = None
        self.dW = None
        self.db = None

    def forward(self, x: np.ndarray) -> np.ndarray:

        self.x = x
        y = (x @ self.weights) + self.bias

        return y

    def backward(self, dy: np.ndarray):

        self.dW = dy.T @ self.x
        self.db = np.sum(dy, axis = 0)

        return self.dW, self.db

```

```python
def loss_fn(y_pred: np.ndarray, y_truth: np.ndarray) -> float:

    loss = np.mean((y_pred - y_truth)**2)

    return loss
```

```python
def derivative(y_pred: np.ndarray, y_truth: np.ndarray):
    
    dy = (2/y_pred.size)*(y_pred - y_truth)

    return dy
```

```python
class Optimizer():
    def __init__(self, model, lr):

        self.model = model
        self.lr = lr
    
    def calc(self):

        self.model.weights -= self.lr * self.model.dW
        self.model.bias -= self.lr * self.model.db
```

```python
X_train = np.random.rand(100, 1)
Y_train = np.random.rand(100, 1) * 2

epochs = 10000
lr = 0.01

model = linear(in_feature = 1, out_feature = 1)
opt = Optimizer(model, lr)

for epoch in range(epochs):

    y_pred = model.forward(X_train)

    loss = loss_fn(y_pred, Y_train)

    dy = derivative(y_pred, Y_train)

    model.backward(dy)

    opt.calc()

    print(f"Epoch: {epoch} | Loss: {loss}")
```
